### a) Informar el total vendido por producto.
Cada fila de la matriz va con un producto. Para el total de un producto, hay que sumar todos los elmentos de esa fila con sum()


```python
codigos = ("P101", "P205", "P330")
ventas_semanales = [
    [12, 15, 10, 18],
    [8, 11, 9, 14],
    [20, 17, 22, 19]
]


def totalPorProducto(codigos, ventas):
    for i in range(len(codigos)):
        total = sum(ventas[i])
        print(f"{codigos[i]}: {total} unidades vendidas")

totalPorProducto(codigos, ventas_semanales)
```

### B) Informar el total de una semana indicada.
Aca seria al reves hayq eu recorrer cada fila (producto) y tomar solo la columna correspondiente a esa semana sumando esos valores 


```python
def totalPorSemana(ventas, semana):
    total = 0
    for producto in ventas:
        total += producto[semana]
    return total

semanaElegida = 2  
resultado = totalPorSemana(ventas_semanales, semanaElegida)
print(f"Total vendido en la semana {semanaElegida + 1}: {resultado} unidades")
```

### c) Determinar el código del producto con mayor venta acumulada.
Se recorre cada producto, se calcula su total con sum() y se va comprando para quedarse con el mayor.

```python
def productoMasVendido(codigos, ventas):
    mejor_codigo = None #Buena practica por si se pasa una lista vacia devuelve none
    mejor_total = -1 #Necesito un valor mas chico que cualquiera venta 
    
    for i in range(len(codigos)):
        total = sum(ventas[i])
        if total > mejor_total:
            mejor_total = total
            mejor_codigo = codigos[i]
    
    return mejor_codigo, mejor_total

codigo, total = productoMasVendido(codigos, ventas_semanales)
print(f"El producto con más ventas es {codigo}, con {total} unidades acumuladas")
```
