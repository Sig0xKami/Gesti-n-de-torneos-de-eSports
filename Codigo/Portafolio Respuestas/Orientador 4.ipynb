## 1. Repaso de las clases anteriores

* **a) ¿Qué significa que un objeto sea mutable o inmutable?**  
  Un objeto mutable puede modificarse después de haber sido creado. Por ejemplo, una lista permite agregar, eliminar o cambiar elementos.  
  Un objeto inmutable no puede modificarse después de haber sido creado. Por ejemplo, una tupla no puede agregar, cambiar, ni eliminar directamente sus elementos.

## 2. Primer contacto con las tuplas

* **a) ¿En qué se parecen una lista y una tupla?**  
  Ambas son estructuras ordenadas que pueden almacenar varios elementos de distintos tipos, se pueden recorrer con un bucle `for`, admiten el uso de rebanadas (*slicing*) y permiten acceder a sus elementos mediante índices numéricos entre corchetes.

* **b) ¿Cuál es la diferencia fundamental entre ambas?**  
  La diferencia principal entre tupla y lista es que las listas son mutables (se pueden modificar sus datos) mientras que las tuplas son inmutables (su contenido y tamaño quedan fijos al crearse).

* **c) ¿Qué excepción produce intentar asignar un valor a una posición de una tupla?**  
  Intentar modificar una tupla produce una excepción de tipo `TypeError` porque no se puede modificar ni reasignar la posición de una tupla (`'tuple' object does not support item assignment`).

* **d) Mencionen un ejemplo de información que convendría representar mediante una tupla:**  
  Un ejemplo apropiado para representar mediante una tupla son los datos de estructura fija que no deben cambiar durante el programa, como los datos de un alumno en un registro escolar: `(1052, "Ana Pérez", 8.5)`.

## 3. Creación de tuplas

En Python, la coma es la que determina la creación de una tupla; los paréntesis mejoran la legibilidad. Por eso, una tupla de un único elemento necesita obligatoriamente una coma al final.

* `type(("Python"))`: Retorna `<class 'str'>` porque sin la coma los paréntesis actúan simplemente como signos de agrupación matemática o de prioridad en una cadena de caracteres.
* `type(("Python",))`: Retorna `<class 'tuple'>` porque la coma le indica al intérprete que se trata de una tupla de un solo elemento.

***
### Ejercicio 3 (Python):

````python
def main():
    diasHabiles = ("Lunes", "Martes", "Miércoles", "Jueves", "Viernes")
    tuplaVacia = ()
    unElemento = ("Python",)
    registro = (1052, "Ana Pérez", 8.5)

    print("diasHabiles:", diasHabiles, type(diasHabiles))
    print("tuplaVacia:", tuplaVacia, type(tuplaVacia))
    print("unElemento:", unElemento, type(unElemento))
    print("registro:", registro, type(registro))

    # Comprobación de la coma
    tipoCadena = type(("Python"))
    tipoTupla = type(("Python",))
    print("\ntype(('Python')) ->", tipoCadena)
    print("type(('Python',)) ->", tipoTupla)


main()
````

## 4. Acceso, recorrido y rebanadas

Dada la tupla: `productos = ("Teclado", "Mouse", "Monitor", "Auriculares", "Webcam")`

* **a)** Primer elemento (`productos[0]`), último elemento (`productos[-1]`) y elemento central (`productos[2]`).
* **b)** Tres primeros elementos mediante slicing: `productos[:3]`.
* **c)** Elementos desde "Monitor" hasta el final: `productos[2:]`.
* **d)** Tupla con el orden invertido: `productos[::-1]`.
* **e)** Recorrido con bucle `for` para mostrar cada elemento.
* **f) Intento de asignación `productos[0] = "Notebook"`:**  
  Produce un error de tipo `TypeError`. Ocurre porque las tuplas al ser inmutables no admiten reasignar ni reemplazar sus valores por posición.  
  Las rebanadas como `productos[:3]` o `productos[::-1]` generan **nuevas tuplas** independientes en memoria y no alteran la tupla original.

***
### Ejercicio 4 (Python):

````python
def main():
    productos = ("Teclado", "Mouse", "Monitor", "Auriculares", "Webcam")

    # a) Primer, ultimo y central
    print("Primer elemento:", productos[0])
    print("Ultimo elemento:", productos[-1])
    print("Elemento central:", productos[2])

    # b) Tres primeros
    print("\nTres primeros elementos:", productos[:3])

    # c) Desde 'Monitor' hasta el final
    print("Desde Monitor hasta el final:", productos[2:])

    # d) Tupla invertida
    invertida = productos[::-1]
    print("Tupla invertida:", invertida)

    # e) Recorrido con for
    print("\n--- Listado de productos ---")
    for producto in productos:
        print("Producto:", producto)

    # f) Verificacion de inmutabilidad
    try:
        productos[0] = "Notebook"
    except TypeError as error:
        print(f"\nError capturado al intentar modificar productos[0]: {error}")


main()
````

## 5. Operadores, funciones y métodos

Dada la tupla: `ventas = (120, 85, 230, 150, 90, 150)`

* **a) Funciones estadísticas:** `len()` para la cantidad total, `max()` para el mayor valor, `min()` para el menor valor y `sum()` para la suma total.
* **b) Pertenencia y ausencia:** Se evalúan mediante `150 in ventas` y `500 not in ventas`.
* **c) Contar apariciones:** Se utiliza `ventas.count(150)`.
* **d) Primera posición:** Se utiliza `ventas.index(230)`.
* **e) Concatenación:** `ventas + (300, 250)` devuelve una nueva tupla combinada sin modificar la original.
* **f) Replicación:** `(0, 1) * 3` genera una nueva tupla repitiendo la secuencia tres veces: `(0, 1, 0, 1, 0, 1)`.
* **Validación previa a `index()`:** Como buscar un elemento inexistente con `.index()` genera un error `ValueError`, se realiza una validación previa con `in`.

***
### Ejercicio 5 (Python):

````python
def main():
    ventas = (120, 85, 230, 150, 90, 150)

    # a) Cantidad, mayor, menor y suma total
    cantidad = len(ventas)
    mayor = max(ventas)
    menor = min(ventas)
    sumaTotal = sum(ventas)

    print("Cantidad de ventas:", cantidad)
    print("Mayor venta:", mayor)
    print("Menor venta:", menor)
    print("Suma total:", sumaTotal)

    # b) Pertenencia y ausencia
    esta150 = 150 in ventas
    noEsta500 = 500 not in ventas
    print("\n¿150 esta en ventas?:", esta150)
    print("¿500 no esta en ventas?:", noEsta500)

    # c) Cantidad de apariciones de 150
    cVeces150 = ventas.count(150)
    print("Apariciones de 150:", cVeces150)

    # d) Primera posicion de 230
    posicion230 = ventas.index(230)
    print("Posicion de 230:", posicion230)

    # e) Concatenacion
    nuevaTupla = ventas + (300, 250)
    print("Tupla concatenada:", nuevaTupla)
    print("¿Es una nueva tupla?:", type(nuevaTupla) is tuple)

    # f) Replicacion
    replicada = (0, 1) * 3
    print("Tupla replicada:", replicada)

    # Validacion previa para evitar ValueError
    buscando = 999
    if buscando in ventas:
        posicion999 = ventas.index(buscando)
        print(f"{buscando} esta en la posicion {posicion999}")
    else:
        print(f"Validacion: {buscando} no esta presente en la tupla.")


main()
````

## 6. Empaquetado y desempaquetado

* **a) ¿Por qué `fecha` es una tupla aun sin paréntesis?**  
  Lo que define una tupla en Python no son los paréntesis sino la presencia de la coma. Al escribir `dia, mes, anio` (valores separados por comas sin delimitadores como corchetes o llaves), el intérprete los empaqueta automáticamente en una tupla (*tuple packing*). Los paréntesis son una ayuda visual opcional.

* **b) Variables obtenidas mediante desempaquetado:**  
  Al ejecutar `diaN, mesN, anioN = fecha`, Python reparte ordenadamente cada elemento de la tupla en cada variable (*tuple unpacking*).

* **c) Desempaquetado en dos variables y análisis del `ValueError`:**  
  Al intentar hacer `a, b = fecha` se produce un error `ValueError: too many values to unpack (expected 2, got 3)`. Ocurre porque la tupla contiene tres elementos y solo se le ofrecieron dos variables para almacenar los datos.

* **d) Condición obligatoria para desempaquetar:**  
  La cantidad de variables del lado izquierdo de la asignación debe ser exactamente igual a la cantidad de elementos que contiene el iterable del lado derecho.

***
### Ejercicio 6 (Python):

````python
def main():
    dia = 25
    mes = "Septiembre"
    anio = 2026
    fecha = dia, mes, anio

    # a) Tipo de dato
    print("Tipo de fecha empaquetada:", type(fecha))

    # b) Desempaquetado
    diaN, mesN, anioN = fecha
    print("diaN:", diaN)
    print("mesN:", mesN)
    print("anioN:", anioN)

    # c) Intento de desempaquetado incorrecto
    try:
        a, b = fecha
    except ValueError as error:
        print(f"Error capturado al intentar desempaquetar: {error}")


main()
````

## 8. Integración con matrices

Integración entre una tupla de códigos fija (`codigos = ("P101", "P205", "P330")`) y una matriz de ventas de 3 filas por 4 columnas (`ventasSemanales`). Cada fila representa las ventas semanales del producto ubicado en el mismo índice de la tupla.

***
### Ejercicio 8 (Python):

````python
def totalPorProducto(codigos, ventas):
    """a) Informa el total vendido por cada producto."""
    print("\n--- Total vendido por producto ---")
    for i in range(len(codigos)):
        total = sum(ventas[i])
        print(f"{codigos[i]}: {total} unidades vendidas")


def totalPorSemana(ventas, semana):
    """b) Retorna el total acumulado de una semana indicada (indice de columna)."""
    total = 0
    for fila in range(len(ventas)):
        total = total + ventas[fila][semana]
    return total


def productoMasVendido(codigos, ventas):
    """c) Determina el codigo del producto con mayor venta acumulada."""
    mejorCodigo = codigos[0]
    mejorTotal = sum(ventas[0])

    for i in range(1, len(codigos)):
        total = sum(ventas[i])
        if total > mejorTotal:
            mejorTotal = total
            mejorCodigo = codigos[i]
            
    return mejorCodigo, mejorTotal


def pedirSemanaValida(cantSemanas):
    """d) Valida la semana ingresada por el usuario manejando posibles errores."""
    semanaValida = -1
    while semanaValida == -1:
        entrada = input(f"Ingrese la semana a consultar (1 a {cantSemanas}): ")
        try:
            numero = int(entrada)
            if 1 <= numero <= cantSemanas:
                semanaValida = numero - 1
            else:
                print(f"Error: La semana debe estar entre 1 y {cantSemanas}.")
        except ValueError:
            print("Error: Debe ingresar un número entero válido.")
    return semanaValida


def main():
    codigos = ("P101", "P205", "P330")
    ventasSemanales = [
        [12, 15, 10, 18],
        [8, 11, 9, 14],
        [20, 17, 22, 19]
    ]

    # a) Total por producto
    totalPorProducto(codigos, ventasSemanales)

    # b y d) Consulta con semana validada
    semanaElegida = pedirSemanaValida(len(ventasSemanales[0]))
    resultadoSemana = totalPorSemana(ventasSemanales, semanaElegida)
    print(f"Total vendido en la semana {semanaElegida + 1}: {resultadoSemana} unidades")

    # c) Producto con mayor venta
    codigoMax, totalMax = productoMasVendido(codigos, ventasSemanales)
    print(f"\nEl producto con más ventas acumuladas es {codigoMax} con {totalMax} unidades.")


main()
````

## 9. Desafío integrador: catálogo de productos

Desarrollo modular de un sistema de carga y consulta de productos. Cada producto se almacena como una tupla inmutable de tres elementos `(codigo, descripcion, precio)` y el catálogo general se administra en una lista mutable. La carga finaliza con `"FIN"`. Si se ingresa un código ya existente, se actualiza el precio reemplazando la tupla completa por una nueva en esa misma posición.

````python
def buscarProducto(productos, codigo):
    """Busca un producto por codigo y retorna su tupla o None si no existe."""
    for i in range(len(productos)):
        if productos[i][0].upper() == codigo.upper():
            return productos[i]
    return None


def buscarIndiceProducto(productos, codigo):
    """Retorna el indice donde se encuentra el producto o -1."""
    for i in range(len(productos)):
        if productos[i][0].upper() == codigo.upper():
            return i
    return -1


def cargarProductos():
    """Carga productos en una lista de tuplas hasta que el codigo sea FIN."""
    catalogo = []
    print("=== Carga del Catálogo de Productos (FIN para terminar) ===")

    while True:
        codigo = input("\nIngrese código del producto: ").strip()
        if codigo.upper() == "FIN":
            break
        if codigo == "":
            print("El código no puede estar vacío.")
            continue

        indiceExistente = buscarIndiceProducto(catalogo, codigo)

        if indiceExistente != -1:
            print(f"El producto con código '{codigo}' ya existe en el catálogo.")
            opcion = input("¿Desea actualizar su precio? (S/N): ").strip().upper()
            if opcion == "S":
                nuevoPrecio = -1.0
                while nuevoPrecio <= 0:
                    try:
                        nuevoPrecio = float(input("Ingrese nuevo precio unitario (> 0): "))
                        if nuevoPrecio <= 0:
                            print("El precio debe ser mayor a cero.")
                    except ValueError:
                        print("Error: Ingrese un valor numérico para el precio.")
                
                # Reemplazo de la tupla completa en la lista
                descripcionActual = catalogo[indiceExistente][1]
                codigoActual = catalogo[indiceExistente][0]
                catalogo[indiceExistente] = (codigoActual, descripcionActual, nuevoPrecio)
                print(f"Precio actualizado para '{codigoActual}'.")
        else:
            descripcion = input("Ingrese descripción: ").strip()
            precio = -1.0
            while precio <= 0:
                try:
                    precio = float(input("Ingrese precio unitario (> 0): "))
                    if precio <= 0:
                        print("El precio debe ser mayor a cero.")
                except ValueError:
                    print("Error: Ingrese un valor numérico para el precio.")
            
            nuevoProducto = (codigo, descripcion, precio)
            catalogo.append(nuevoProducto)
            print(f"Producto '{codigo}' agregado correctamente.")

    return catalogo


def mostrarProductos(productos):
    """Muestra el catalogo completo de productos."""
    if len(productos) == 0:
        print("\nEl catálogo está vacío.")
    else:
        print("\n--- Catálogo de Productos ---")
        for i in range(len(productos)):
            cod, desc, prec = productos[i]
            print(f"Item {i + 1}: [{cod}] {desc} - ${prec:.2f}")


def productoMayorPrecio(productos):
    """Retorna la tupla del producto con mayor precio o None si la lista esta vacia."""
    if len(productos) == 0:
        return None
    
    mayorProducto = productos[0]
    for i in range(1, len(productos)):
        if productos[i][2] > mayorProducto[2]:
            mayorProducto = productos[i]
    return mayorProducto


def precioPromedio(productos):
    """Retorna el promedio de precios o None si no hay productos."""
    if len(productos) == 0:
        return None
    
    sumaPrecios = 0
    for i in range(len(productos)):
        sumaPrecios = sumaPrecios + productos[i][2]
    return sumaPrecios / len(productos)


def main():
    catalogo = cargarProductos()
    mostrarProductos(catalogo)

    if len(catalogo) > 0:
        masCaro = productoMayorPrecio(catalogo)
        print(f"\nProducto más caro: [{masCaro[0]}] {masCaro[1]} a ${masCaro[2]:.2f}")

        promedio = precioPromedio(catalogo)
        print(f"Precio promedio del catálogo: ${promedio:.2f}")

        # Búsqueda interactiva
        codigoBuscado = input("\nIngrese código a buscar: ").strip()
        encontrado = buscarProducto(catalogo, codigoBuscado)
        if encontrado is not None:
            print(f"Encontrado: [{encontrado[0]}] {encontrado[1]} - ${encontrado[2]:.2f}")
        else:
            print(f"El producto '{codigoBuscado}' no existe en el catálogo.")
    else:
        print("\nNo se cargaron productos para analizar.")


main()
````
