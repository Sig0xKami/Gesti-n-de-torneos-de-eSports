## 1. Finalización anticipada de un ciclo: break

La instrucción `break` detiene de forma inmediata la ejecución del bucle `for` o `while` más cercano en el que se encuentra. La ejecución continúa en la primera línea que esté después de ese ciclo.

### Análisis de preguntas teóricas:
* **b) ¿En qué caso se ejecuta `break`?:** Se ejecuta en el momento exacto en que la operación `numero % i == 0` da verdadero, es decir, apenas el programa encuentra un número que divide exactamente al valor ingresado (distinto de 1 y de sí mismo).
* **c) ¿Por qué evita iteraciones innecesarias?:** Porque una vez encontrado un único divisor ya tenemos la certeza matemática de que el número no es primo. Seguir iterando hasta llegar a `numero - 1` sería un desperdicio de procesamiento sin sentido alguno.
* **d) Comparación con una versión que recorre todos los divisores:** La versión sin `break` tiene que recorrer obligatoriamente todo el rango desde 2 hasta `n - 1`, mientras que con `break`, en números compuestos como un número par grande, se detiene en la primera vuelta (`i = 2`), ahorrando cientos o miles de vueltas.

***
### Ejercicio 1 (Python):

```python
def esPrimo(numero):
    """Determina si un numero es primo utilizando variable logica y break."""
    if numero < 2:
        return False

    esNumeroPrimo = True
    for i in range(2, numero):
        if numero % i == 0:
            esNumeroPrimo = False
            break

    return esNumeroPrimo


def main():
    numerosPrueba = [-3, 0, 1, 2, 9, 17, 25]

    print(" Prueba de la funcion numero primo ")
    for n in numerosPrueba:
        resultado = esPrimo(n)
        print(f"El numero {n} es primo: {resultado}")


main()
```

## 2. Ciclos controlados con while True

Un bucle `while True` genera una repetición cuya condición siempre resulta verdadera. Se utiliza cuando no sabemos de antemano cuántos datos va a ingresar el usuario y la condición de corte se evalúa adentro del bloque para ejecutar un `break`.

En este ejercicio se cargan números enteros hasta que se ingresa `-1`. Ese valor es solo una señal de fin (centinela), por lo que se debe descartar inmediatamente y no sumarlo a los acumuladores ni contadores.

***
### Ejercicio 2 (Python):

```python
def main():
    cantidad = 0
    sumaTotal = 0

    print("=== Ingreso de numeros (ingrese -1 para terminar) ===")
    while True:
        numero = int(input("Ingrese un numero entero: "))
        if numero == -1:
            break
        
        sumaTotal = sumaTotal + numero
        cantidad = cantidad + 1

    print("\n--- Resultados ---")
    if cantidad == 0:
        print("No existen datos para calcular el promedio (se ingreso -1 al inicio).")
    else:
        promedio = round(sumaTotal / cantidad, 2)
        print(f"Cantidad de numeros ingresados: {cantidad}")
        print(f"Suma total: {sumaTotal}")
        print(f"Promedio: {promedio}")


main()
```

## 3. Concepto de excepción

Una excepción es un error que ocurre durante la ejecución de un programa cuando la sintaxis está bien pero una operación no se puede realizar. Si no se captura, el programa se interrumpe y muestra un traceback en la terminal.

### Casos analizados sobre `resultado = 100 / numero`:
* **Con entrada `10`:** El programa convierte `10` a entero y calcula `100 / 10 = 10.0` sin problemas.
* **Con entrada `0`:** Falla en la división matemática y lanza un error de tipo `ZeroDivisionError: division by zero`.
* **Con entrada `"hola"`:** Falla al intentar convertir el texto a número con `int()` y lanza un `ValueError: invalid literal for int() with base 10: 'hola'`.

### Validación preventiva (`if`) vs. Captura correctiva (`try-except`):
* **Validación preventiva (`if`):** Verifica las condiciones antes de que la instrucción se ejecute (por ejemplo: `if numero != 0:`). Es ideal para reglas de negocio predecibles.
* **Captura correctiva (`try-except`):** Intenta ejecutar el bloque de código y, si ocurre un fallo no previsto o incontrolable (como que el usuario meta letras cuando esperamos números), captura la excepción en el momento y evita que el programa se cuelgue.

***
### Ejercicio 3 (Python):

```python
def main():
    print("--- Demostracion de excepciones ---")
    entrada = input("Ingrese un numero para dividir a 100: ")

    try:
        numero = int(entrada)
        resultado = 100 / numero
        print(f"El resultado es: {resultado}")
    except ZeroDivisionError:
        print("Error: No se puede dividir por cero (ZeroDivisionError).")
    except ValueError:
        print("Error: Debe ingresar un numero entero valido, no texto (ValueError).")


main()
```

## 4. Captura con try-except

El bloque `try` debe contener únicamente las instrucciones que pueden fallar. Si se captura la excepción específica `ValueError`, se le puede avisar al usuario que cometió un error de tipeo sin perder los datos que ya venía cargando en la sesión.

***
### Ejercicio 4 (Python):

```python
def main():
    numerosValidos = []

    print("=== Carga de 3 numeros validos ===")
    while len(numerosValidos) < 3:
        entrada = input(f"Ingrese el numero {len(numerosValidos) + 1}: ")
        try:
            valor = int(entrada)
            numerosValidos.append(valor)
        except ValueError:
            print("Error: El dato ingresado no es un numero entero. Intente nuevamente.")

    print(f"\nLista de numeros cargados correctamente: {numerosValidos}")


main()
```

## 5. Múltiples excepciones y reglas del problema

Dada la lista `numeros = [10, 20, 30, 40]`:

En Python estándar, los índices negativos son válidos para leer desde el final (por ejemplo, `numeros[-1]` devuelve `40`). Sin embargo, si la regla del problema exige estrictamente posiciones entre `0` y `len(numeros) - 1`, debemos validar el caso negativo explícitamente y provocar un `ValueError` con la instrucción `raise`.

### Pruebas realizadas:
* `"dos"` -> Provoca `ValueError` en `int()`.
* `1` -> Caso válido, devuelve `20`.
* `8` -> Provoca `IndexError` porque la lista solo tiene índices de 0 a 3.
* `-1` -> Provoca `ValueError` forzado por nuestra regla de negocio.

**¿Por qué conviene mostrar un mensaje diferente para cada problema?:**  
Porque ayuda al usuario a saber qué hizo mal: no es lo mismo equivocarse al tipear letras cuando se pide un número que ingresar un índice fuera de rango. Mensajes específicos facilitan la corrección y la depuración del código.

***
### Ejercicio 5 (Python):

```python
def main():
    numeros = [10, 20, 30, 40]
    print(f"Lista disponible: {numeros} (posiciones validas de 0 a {len(numeros) - 1})")

    entrada = input("Ingrese una posicion para consultar: ")
    try:
        posicion = int(entrada)
        if posicion < 0:
            raise ValueError("No se permiten indices negativos segun las reglas del problema.")
        
        valor = numeros[posicion]
        print(f"El valor en la posicion {posicion} es: {valor}")

    except ValueError as error:
        print(f"Error de valor: {error}")
    except IndexError:
        print(f"Error de rango: La posicion ingresada no existe en la lista (IndexError).")


main()
```

## 6. Cláusulas else y finally

* **`except`:** Atrapa y gestiona un tipo específico de excepción si ocurrió un fallo en el bloque `try`.
* **`else`:** Se ejecuta única y exclusivamente si el bloque `try` finalizó sin ninguna excepción. Sirve para poner instrucciones que dependen de que el cálculo haya salido bien.
* **`finally`:** Se ejecuta **siempre**, tanto si el bloque funcionó bien como si saltó un error. Se usa para tareas de cierre garantizadas (como cerrar un archivo o mostrar que el intento terminó).

***
### Ejercicio 6 (Python):

```python
def main():
    print("=== Division segura de dos numeros ===")
    entradaA = input("Ingrese el dividendo: ")
    entradaB = input("Ingrese el divisor: ")

    try:
        numeroA = float(entradaA)
        numeroB = float(entradaB)
        resultado = numeroA / numeroB
    except ValueError:
        print("Error: Debe ingresar valores numericos.")
    except ZeroDivisionError:
        print("Error: No es posible dividir por cero.")
    else:
        print(f"La division se realizo con exito. Resultado: {round(resultado, 2)}")
    finally:
        print("Operacion finalizada (el bloque finally siempre se ejecuta).")


main()
```

## 7. Provocar una excepción con raise

La instrucción `raise` permite forzar una excepción en nuestro código cuando un dato no cumple con las reglas del negocio, aun cuando el tipo de dato sea correcto. 

En este caso, una cantidad de `-5` o un precio de `$0` son valores numéricos válidos para Python, pero representan una incoherencia para un sistema comercial. Por eso lanzamos `raise ValueError` con un mensaje explicativo.

***
### Ejercicio 7 (Python):

```python
def calcularImporte(cantidad, precio):
    """Calcula el importe total validando que cantidad y precio sean mayores a cero."""
    if cantidad <= 0:
        raise ValueError("La cantidad debe ser mayor a cero.")
    if precio <= 0:
        raise ValueError("El precio unitario debe ser mayor a cero.")

    return cantidad * precio


def main():
    print("--- Prueba de calcularImporte con raise ---")
    pruebas = [
        (3, 1200),
        (0, 500),
        (4, -200)
    ]

    for cant, prec in pruebas:
        try:
            total = calcularImporte(cant, prec)
            print(f"Compra valida -> Cantidad: {cant}, Precio: ${prec} | Total: ${total}")
        except ValueError as error:
            print(f"Error detectado con ({cant}, {prec}) -> {error}")


main()
```

## 8. Comprobaciones internas con assert

La instrucción `assert` comprueba una condición que nosotros como programadores sabemos que siempre debe cumplirse. Si la condición resulta falsa, Python dispara un `AssertionError`.

**Importante:** `assert` es una herramienta pensada para depurar y encontrar errores de lógica interna del código, nunca para validar los datos que ingresa el usuario (ya que las aserciones pueden desactivarse en entornos de producción).

***
### Ejercicio 8 (Python):

```python
def calcularImporte(cantidad, precio):
    return cantidad * precio


def main():
    # 1. Comprobacion de la consigna (pasa sin problemas)
    importe = calcularImporte(3, 1200)
    assert importe == 3600, "El importe calculado no es el esperado"
    print("Asercion 1 superada: 3 * 1200 es 3600.")

    # 2. Segunda asercion correcta
    importeDos = calcularImporte(2, 500)
    assert importeDos == 1000, "El calculo de 2 * 500 debe dar 1000"
    print("Asercion 2 superada: 2 * 500 es 1000.")

    # 3. Asercion que falla intencionalmente para ver el error
    try:
        assert importeDos == 9999, "Error de logica interna: el importe esperado era 1000"
    except AssertionError as error:
        print(f"Asercion 3 capturada con exito -> AssertionError: {error}")


main()
```

## 9. Desafío integrador: carga segura de productos

Desarrollo modular de un sistema de registro de productos en una lista en memoria. Cada producto se guarda como una lista `[codigo, descripcion, cantidad, precio]`. La carga se corta cuando se escribe `"FIN"`.

El programa valida datos no numéricos con `try-except`, rechaza valores menores o iguales a cero mediante `raise ValueError`, usa `assert` para verificar la lógica interna del cálculo del total y calcula los promedios evitando divisiones por cero.

***
### Ejercicio 9 (Python):

```python
def calcularImporteProducto(cantidad, precio):
    """Calcula el total de un producto con asercion de control interno."""
    total = cantidad * precio
    assert total >= 0, "El importe calculado nunca puede ser negativo"
    return total


def cargarProductos():
    """Carga productos de forma segura en una lista hasta que el codigo sea FIN."""
    catalogo = []
    print("=== Carga Segura de Productos (escriba FIN para terminar) ===")

    while True:
        codigo = input("\nIngrese codigo del producto: ").strip()
        if codigo.upper() == "FIN":
            break
        if codigo == "":
            print("El codigo no puede estar vacio.")
            continue

        descripcion = input("Ingrese descripcion: ").strip()

        try:
            cantidad = int(input("Ingrese cantidad: "))
            precio = float(input("Ingrese precio unitario: "))

            if cantidad <= 0:
                raise ValueError("La cantidad debe ser mayor a cero.")
            if precio <= 0:
                raise ValueError("El precio debe ser mayor a cero.")

            productoNuevo = [codigo, descripcion, cantidad, precio]
            catalogo.append(productoNuevo)
            print(f"Producto '{codigo}' cargado exitosamente.")

        except ValueError as error:
            print(f"Carga rechazada: {error} Intente nuevamente con este producto.")

    return catalogo


def mostrarResumenCatalogo(catalogo):
    """Muestra el reporte general calculando totales y promedios seguros."""
    print("\n================ RESUMEN DE PRODUCTOS ================")
    cantidadProductos = len(catalogo)

    if cantidadProductos == 0:
        print("No se registraron productos en el catalogo.")
        return

    importeAcumulado = 0.0
    sumaPrecios = 0.0

    print("Cod   | Descripcion          | Cant | Precio Unit | Subtotal")
    print("------+----------------------+------+-------------+----------")
    for i in range(cantidadProductos):
        prod = catalogo[i]
        subtotal = calcularImporteProducto(prod[2], prod[3])
        importeAcumulado = importeAcumulado + subtotal
        sumaPrecios = sumaPrecios + prod[3]

        print(f"{prod[0]:<5} | {prod[1]:<20} | {prod[2]:<4} | ${prod[3]:<10} | ${subtotal}")

    precioPromedio = round(sumaPrecios / cantidadProductos, 2)

    print("------------------------------------------------------")
    print(f"Cantidad total de productos distintos: {cantidadProductos}")
    print(f"Importe total acumulado: ${round(importeAcumulado, 2)}")
    print(f"Precio unitario promedio: ${precioPromedio}")


def main():
    productosCargados = cargarProductos()
    mostrarResumenCatalogo(productosCargados)


main()
```

### Análisis técnico conceptual

* **Diferencias entre `break`, `try-except`, `raise` y `assert`:**
  * **`break`:** Es una sentencia de control de flujo que corta inmediatamente un bucle (`for` o `while`).
  * **`try-except`:** Es una estructura pensada para atrapar errores en tiempo de ejecución y recuperarse de ellos sin que el programa se cuelgue.
  * **`raise`:** Se usa para provocar una excepción a propósito cuando no se cumple una regla de nuestro programa, obligando al código a gestionarla.
  * **`assert`:** Es una comprobación interna para el desarrollador que asegura que una condición matemática o lógica sea verdadera durante la depuración.

* **Justificación de `else` y `finally` en la división:**  
  En el ejercicio de división, el bloque `else` garantiza que el mensaje con el resultado solo se imprima si la cuenta no falló (evitando mostrar resultados vacíos o con error). El bloque `finally` se utilizó para asegurar un mensaje de cierre de operación que se ejecute siempre, haya ocurrido un error o no, simulando el cierre de recursos en memoria.
