## 1. Repaso de la clase anterior

* **a) Diferencia entre problema, algoritmo y programa:**
  * **Problema:** Es la necesidad o situación del mundo real que tenemos que resolver y comprender antes de tocar código.
  * **Algoritmo:** Es la serie ordenada, lógica y finita de pasos que diseñamos para resolver el problema, sin depender de ningún lenguaje.
  * **Programa:** Es la traducción concreta de ese algoritmo a instrucciones que la computadora entiende, usando las reglas y sintaxis de un lenguaje como Python.

* **b) Ventajas de dividir un programa en funciones:**
  * Hace que el código sea mucho más ordenado y fácil de entender.
  * Permite reutilizar código sin tener que copiar y pegar las mismas líneas varias veces.
  * Facilita encontrar y corregir errores rápido, porque si algo falla sabemos a qué función puntual ir.

* **c) Diferencia entre mostrar un resultado y retornarlo:**
  * **Mostrar (`print`):** Solo imprime el texto en la consola para que lo lea el usuario. El valor no queda guardado y el resto del programa no lo puede usar (devuelve `None`).
  * **Retornar (`return`):** Devuelve el resultado calculado al lugar donde llamamos a la función. Ese dato queda en memoria para guardarlo en una variable o seguir haciendo cuentas.

* **d) Acceso, modificación y recorrido de una lista:**
  * **Acceso:** Se usa el índice de la posición entre corchetes, empezando siempre desde cero (por ejemplo, `miLista[0]`).
  * **Modificación:** Se le asigna un valor nuevo a una posición que ya existe (por ejemplo, `miLista[1] = "Nuevo"`).
  * **Recorrido:** Se hace con un bucle `for`, ya sea por elemento (`for elemento in miLista`) o por índice clásico con `range(len(miLista))`.

## 4. Rebanadas (slicing) para obtener nuevas listas

Lista base:
`juegos = ["Minecraft", "Valorant", "Fortnite", "FIFA", "Rocket League", "Roblox", "LOL", "Among Us"]`

* **a) Los tres primeros elementos:** `juegos[:3]` -> `['Minecraft', 'Valorant', 'Fortnite']`
* **b) Los cuatro últimos elementos:** `juegos[-4:]` -> `['Rocket League', 'Roblox', 'LOL', 'Among Us']`
* **c) Desde la posición 1 hasta la 4 inclusive:** `juegos[1:5]` -> `['Valorant', 'Fortnite', 'FIFA', 'Rocket League']`
* **d) Elementos de posiciones pares:** `juegos[::2]` -> `['Minecraft', 'Fortnite', 'Rocket League', 'LOL']`
* **e) Elementos de posiciones impares:** `juegos[1::2]` -> `['Valorant', 'FIFA', 'Roblox', 'Among Us']`
* **f) Lista invertida:** `juegos[::-1]` -> `['Among Us', 'LOL', 'Roblox', 'Rocket League', 'FIFA', 'Fortnite', 'Valorant', 'Minecraft']`
* **g) Todos excepto el primero:** `juegos[1:]` -> `['Valorant', 'Fortnite', 'FIFA', 'Rocket League', 'Roblox', 'LOL', 'Among Us']`
* **h) Todos excepto el último:** `juegos[:-1]` -> `['Minecraft', 'Valorant', 'Fortnite', 'FIFA', 'Rocket League', 'Roblox', 'LOL']`

### Ejercicio 4 (Python):

```python
def main():
    juegos = ["Minecraft", "Valorant", "Fortnite", "FIFA", "Rocket League", "Roblox", "LOL", "Among Us"]

    print("Lista original:", juegos)
    print("a) Tres primeros:", juegos[:3])
    print("b) Cuatro ultimos:", juegos[-4:])
    print("c) Posicion 1 a 4 inclusive:", juegos[1:5])
    print("d) Posiciones pares:", juegos[::2])
    print("e) Posiciones impares:", juegos[1::2])
    print("f) Invertida:", juegos[::-1])
    print("g) Todos menos el primero:", juegos[1:])
    print("h) Todos menos el ultimo:", juegos[:-1])


main()
```

## 5. Modificación de listas mediante slicing

Partiendo de: `numeros = [2, 4, 6, 8, 10, 12, 14]`

* **a) Reemplazar 6 y 8 por 60 y 80:** `copia[2:4] = [60, 80]`
* **b) Eliminar 10 y 12:** `copia[4:6] = []`
* **c) Insertar 100 y 200 entre 4 y 6:** `copia[2:2] = [100, 200]` (rebanada nula en la posición 2)
* **d) Agregar tres valores al comienzo:** `copia[:0] = [-2, -1, 0]`
* **e) Vaciar la lista:** `copia[:] = []`

### Ejercicio 5 (Python):

```python
def main():
    numeros = [2, 4, 6, 8, 10, 12, 14]

    # a) Reemplazar 6 y 8 por 60 y 80
    copiaA = numeros.copy()
    copiaA[2:4] = [60, 80]
    print("a) Reemplazo de 6 y 8:", copiaA)

    # b) Eliminar 10 y 12
    copiaB = numeros.copy()
    copiaB[4:6] = []
    print("b) Eliminar 10 y 12:", copiaB)

    # c) Insertar 100 y 200 entre 4 y 6
    copiaC = numeros.copy()
    copiaC[2:2] = [100, 200]
    print("c) Insertar 100 y 200:", copiaC)

    # d) Agregar tres valores al comienzo
    copiaD = numeros.copy()
    copiaD[:0] = [-2, -1, 0]
    print("d) Agregar tres valores al inicio:", copiaD)

    # e) Vaciar la lista
    copiaE = numeros.copy()
    copiaE[:] = []
    print("e) Lista vaciada:", copiaE)


main()
```

## 6. Listas por comprensión

Partiendo de: `puntajes = [45, 80, 63, 91, 100, 72]`

* **a) Puntajes duplicados (Transformación):** `[p * 2 for p in puntajes]`
* **b) Solamente aprobados de 60 o más (Filtro):** `[p for p in puntajes if p >= 60]`
* **c) Cuadrado de los valores pares (Filtro + Transformación):** `[p ** 2 for p in puntajes if p % 2 == 0]`
* **d) Texto "Aprobado" o "Revisar" (Expresión condicional):** `["Aprobado" if p >= 60 else "Revisar" for p in puntajes]`
* **e) ¿Por qué la lista original no se modifica?:**
  Porque las listas por comprensión crean y retornan una lista nueva en memoria con los resultados obtenidos, sin alterar los elementos ni la estructura de la lista base.

### Ejercicio 6 (Python):

```python
def main():
    puntajes = [45, 80, 63, 91, 100, 72]

    duplicados = [p * 2 for p in puntajes]
    print("a) Puntajes duplicados:", duplicados)

    aprobados = [p for p in puntajes if p >= 60]
    print("b) Solo aprobados:", aprobados)

    cuadradoPares = [p ** 2 for p in puntajes if p % 2 == 0]
    print("c) Cuadrado de los pares:", cuadradoPares)

    estados = ["Aprobado" if p >= 60 else "Revisar" for p in puntajes]
    print("d) Condicion de aprobacion:", estados)

    print("e) Lista original intacta:", puntajes)


main()
```

## 7. Funciones lambda

Definición de funciones cortas de una sola línea:
* `cuadrado`: calcula el cuadrado de un número.
* `esPar`: retorna `True` si es par y `False` si es impar.
* `mayor`: compara dos números y devuelve el más grande.
* `aplicarDescuento`: calcula el precio final descontando un porcentaje.

### Justificación (Punto e)
En un programa real escribiríamos **`aplicarDescuento`** con una función clásica `def`.  
Un cálculo comercial o de precios necesita validaciones obligatorias (por ejemplo verificar que el precio no sea negativo y que el porcentaje esté entre 0 y 100) para evitar errores. Las funciones lambda son para operaciones directas y simples; cuando necesitamos validar o meter más lógica, siempre conviene usar `def`.

### Ejercicio 7 (Python):

```python
def main():
    cuadrado = lambda x: x ** 2
    esPar = lambda x: x % 2 == 0
    mayor = lambda a, b: a if a > b else b
    aplicarDescuento = lambda precio, porcentaje: precio - (precio * porcentaje / 100)

    print("a) Cuadrado de 5:", cuadrado(5))
    print("b) ¿Es par el 8?:", esPar(8))
    print("c) Mayor entre 15 y 42:", mayor(15, 42))
    print("d) Descuento del 10% a $5000:", aplicarDescuento(5000, 10))


main()
```

## 8. Desafío integrador

Integración de las operaciones del catálogo de videojuegos aplicando diseño modular, búsquedas clásicas, slicing, listas por comprensión y funciones lambda.

Se separo el programa completo con dos comentarios debido a que estaban en dos archivos .py distintos.

```python
# Funciones modulo videojuegos.py

def mostrarJuegos(juegos):
    """Muestra todos los juegos del catalogo recorriendo la lista por indice."""
    if len(juegos) == 0:
        print("El catalogo esta vacio.")
    else:
        print("\n--- Catalogo de Videojuegos ---")
        for i in range(len(juegos)):
            print(f"Juego {i + 1}: {juegos[i]}")


def buscarJuego(juegos, titulo):
    """Retorna la posicion del juego o -1 si no existe."""
    for i in range(len(juegos)):
        if juegos[i].lower() == titulo.lower():
            return i
    return -1


def agregarJuego(juegos, titulo):
    """Agrega un juego si no esta repetido. Modifica la lista original."""
    posicion = buscarJuego(juegos, titulo)
    if posicion == -1:
        juegos.append(titulo)
        return True
    return False


def cantidadJuegos(juegos):
    """Retorna la cantidad total de titulos."""
    return len(juegos)


# Programa principal
def main():
    # Funcion lambda para quitar espacios y normalizar texto
    limpiarTexto = lambda t: t.strip()

    # 1. Catalogo inicial de 10 juegos
    catalogo = [
        "Minecraft", "Valorant", "Fortnite", "FIFA", "Rocket League",
        "Roblox", "LOL", "Among Us", "Cyberpunk", "Terraria"
    ]

    print("=== Sistema de Gestion de Videojuegos ===")

    # Mostrar catalogo y cantidad
    mostrarJuegos(catalogo)
    total = cantidadJuegos(catalogo)
    print(f"\nCantidad total de juegos: {total}")

    # 2. Buscar titulo ingresado por usuario
    busqueda = input("\nIngrese el titulo de un juego para buscar: ")
    busquedaLimpia = limpiarTexto(busqueda)
    posicionEncontrada = buscarJuego(catalogo, busquedaLimpia)

    if posicionEncontrada != -1:
        print(f"El juego '{busquedaLimpia}' se encuentra en la posicion {posicionEncontrada}.")
    else:
        print(f"El juego '{busquedaLimpia}' no esta en el catalogo.")

    # 3. Agregar nuevo titulo evitando duplicados
    nuevoJuego = input("\nIngrese un nuevo titulo para agregar: ")
    nuevoJuegoLimpio = limpiarTexto(nuevoJuego)

    sePudoAgregar = agregarJuego(catalogo, nuevoJuegoLimpio)
    if sePudoAgregar:
        print(f"Exito: '{nuevoJuegoLimpio}' fue agregado al catalogo.")
    else:
        print(f"Aviso: '{nuevoJuegoLimpio}' ya existe, no se agrego.")

    # 4. Operaciones con slicing
    print("\n--- Consultas con Slicing ---")
    print("Primeros 5 titulos:", catalogo[:5])
    print("Ultimos 3 titulos:", catalogo[-3:])
    print("Catalogo invertido:", catalogo[::-1])

    # 5. Lista por comprension: titulos con mas de 8 caracteres
    titulosLargos = [juego for juego in catalogo if len(juego) > 8]
    print("\nTitulos con mas de 8 caracteres:", titulosLargos)


main()
```
