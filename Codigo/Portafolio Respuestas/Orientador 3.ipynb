* **a) ¿Qué ventajas ofrece separar un programa en módulos?**  
  Separar un programa en módulos permite organizar mejor el código, facilitar su lectura y mantenimiento, reutilizar funciones y trabajar de manera más ordenada. Además, si aparece un error, es más fácil localizarlo en un módulo específico.

## 2. Git y GitHub: conceptos fundamentales

* **a) Repositorio local:** Es la copia del proyecto que se encuentra en nuestra computadora y donde Git registra los cambios realizados.
* **b) Repositorio remoto:** Es el repositorio alojado en una plataforma como GitHub, que permite almacenar y compartir el proyecto con otros integrantes.
* **c) Área de preparación o staging:** Es el espacio donde se colocan los archivos o cambios que queremos incluir en el próximo commit mediante `git add`.
* **d) Commit:** Es el registro de un conjunto de cambios realizados en el proyecto, acompañado de un mensaje descriptivo.
* **e) Push y pull:** `git push` envía los cambios del repositorio local al repositorio remoto. `git pull` descarga e integra los cambios del repositorio remoto en la copia local.

## 5. Cadenas de caracteres

Dada la cadena: `texto = "Programacion en Python"`

* **a)** Primer carácter (`texto[0]`) y último carácter (`texto[-1]`).
* **b)** Palabra "Programacion" obtenida mediante slicing (`texto[:12]`).
* **c)** Cadena invertida obtenida mediante slicing (`texto[::-1]`).
* **d)** Verificación de pertenencia de la palabra "Python" en el texto.
* **e) Intento de modificación `texto[0] = "p"`:**  
  Genera un error de tipo `TypeError: 'str' object does not support item assignment`.  
  *Explicación:* Las cadenas de texto (`str`) son un tipo de dato inmutable. Esto es porque una vez que se crea la cadena en la memoria de la computadora, su contenido queda fijo y no se puede alterar posición por posición.

***
### Ejercicio 5 (Python):

````python
def main():
    texto = "Programacion en Python"

    # a)
    print(texto[0])
    print(texto[-1])

    # b)
    palabra = texto[:12]
    print(palabra)

    # c) Mostrar cadena invertida
    invertida = texto[::-1]
    print(invertida)

    # d)
    if "Python" in texto:
        print("Se encuentra la palabra Python en la cadena")
    else:
        print("No se encuentra la cadena python en la cadena")

    # e) Demostracion del error por inmutabilidad
    try:
        texto[0] = "p"
    except TypeError as error:
        print(f"Error obtenido al intentar modificar: {error}")


main()
````

## 6. Funciones y métodos de cadenas

Resultados de las operaciones sobre `texto = "Programacion en Python"`:
* `texto.upper()` -> `"PROGRAMACION EN PYTHON"`
* `texto.lower()` -> `"programacion en python"`
* `texto.title()` -> `"Programacion En Python"`
* `texto.capitalize()` -> `"Programacion en python"`
* `texto.replace("Python", "IA")` -> `"Programacion en IA"`
* `texto.count("a")` -> `2`
* `texto.find("Python")` -> `16`
* `len(texto)` -> `22`

### Métodos de validación

* `isdigit()`: Devuelve `True` solamente si todos los caracteres son dígitos, no tiene ningún signo especial y al menos tiene un carácter.
* `isalnum()`: Devuelve `True` solamente si todos los caracteres son letras o números. Si hay un espacio o símbolo, devuelve `False`.
* `isalpha()`: Retorna `False` en `"Programacion en Python"` porque el método evalúa que absolutamente TODOS los caracteres de la cadena sean letras alfabéticas. Como los espacios en blanco (`" "`) se consideran caracteres dentro de la secuencia y no forman parte del abecedario, la condición falla.

***
### Ejercicio 6 (Python):   

````python
def main():
    texto = "Programacion en Python"

    print("--- Operaciones sobre texto ---")
    print(texto.upper())
    print(texto.lower())
    print(texto.title())
    print(texto.capitalize())
    print(texto.replace("Python", "IA"))
    print(texto.count("a"))
    print(texto.find("Python"))
    print(len(texto))

    print("\n--- Métodos de validación ---")
    c1 = "Programacion"
    c2 = "2026"
    c3 = "Python3"
    c4 = "Programacion en Python"

    print(f"'{c1}' -> isalpha: {c1.isalpha()} | isdigit: {c1.isdigit()} | isalnum: {c1.isalnum()}")
    print(f"'{c2}' -> isalpha: {c2.isalpha()} | isdigit: {c2.isdigit()} | isalnum: {c2.isalnum()}")
    print(f"'{c3}' -> isalpha: {c3.isalpha()} | isdigit: {c3.isdigit()} | isalnum: {c3.isalnum()}")
    print(f"'{c4}' -> isalpha: {c4.isalpha()} | isdigit: {c4.isdigit()} | isalnum: {c4.isalnum()}")


main()
````

## 7. Conversión de tipos y formateo

### Comparación entre f-strings y concatenación tradicional

* **Legibilidad:** Las f-strings son mucho más claras y directas de leer porque permiten colocar las variables entre llaves `{}` dentro del mismo mensaje sin interrumpir la frase.
* **Cantidad de conversiones necesarias:** Con concatenación tradicional es obligatorio convertir cada número usando `str()` manualmente; si nos olvidamos de uno, el programa se rompe con un error de tipos. Con f-strings, Python realiza la conversión a texto de manera automática.
* **Formateo numérico:** La f-string permite aplicar fácilmente el formato de dos decimales usando `{total:.2f}`, mientras que con concatenación hay que recurrir a la función `round(total, 2)` antes de pasarlo a `str()`.

***
### Ejercicio 7 (Python):

````python
def main():
    producto = input("Producto: ")
    precio = float(input("Precio unitario: "))
    cantidad = int(input("Cantidad: "))

    total = precio * cantidad

    # a) Utilizando f-string con formato de dos decimales
    print(f"\nSalida con f-string: Producto: {producto} | Cantidad: {cantidad} | Total: ${total:.2f}")

    # b) Utilizando concatenacion y conversion explicita con str()
    totalRedondeado = round(total, 2)
    mensajeConcatenado = "Salida con concatenacion: Producto: " + producto + " | Cantidad: " + str(cantidad) + " | Total: $" + str(totalRedondeado)
    print(mensajeConcatenado)


main()
````

## 8. Actividad integradora: perfil del equipo

Implementación del archivo `codigo/perfil_equipo.py` para registrar y procesar la información del equipo, sus integrantes y sus roles iniciales.

***
### Ejercicio 8 (Python):

````python
def recorrerListaIntegrantes(integrantesLista, nombre):
    for i in range(len(integrantesLista)):
        if integrantesLista[i] == nombre:
            return nombre
    return -1


def recorrerListaRol(rolLista, nombre, integrantesLista):
    nombreEncontrado = recorrerListaIntegrantes(integrantesLista, nombre)
    if nombreEncontrado == -1: 
        return False
    posicion = integrantesLista.index(nombre)
    rol = rolLista[posicion]
    return rol


def todoRol(rolLista):
    for i in range(len(rolLista)):
        print(f"ROL {rolLista[i]} del integrante")


def todoIntegrante(integrantesLista):
    for i in range(len(integrantesLista)):
        print(f"INTEGRANTE {integrantesLista[i]}")


def asignarIntegrante(nombreIntegrante, listaIntegrantes):
    if recorrerListaIntegrantes(listaIntegrantes, nombreIntegrante) == -1: 
        listaIntegrantes.append(nombreIntegrante)
        return f"Se agrego satifactoriamente al integrante {nombreIntegrante}"
    else:
        return f"El integrante {nombreIntegrante} ya existe, por lo cual no se agrego a la lista. "


def asignarRol(rolLista, nombre, integrantesLista, rol, nombreEquipo): 
    if recorrerListaIntegrantes(integrantesLista, nombre) == -1:
        return "El integrante no se encuentra en la lista"
    else:
        rolLista.append(rol)
        return f"El integrante {nombre} del equipo {nombreEquipo} fue asignado con el ROL {rol} "


def procesarAltaIntegrante(nombreIntegrante, listaIntegrantes, rolLista, rolIntegrante, nombreEquipo):
    if recorrerListaIntegrantes(listaIntegrantes, nombreIntegrante) != -1:
        return f"El integrante {nombreIntegrante} ya existe, no se agregó ni se le asignó rol."
    else:
        listaIntegrantes.append(nombreIntegrante)
        rolLista.append(rolIntegrante)
        return f"El integrante {nombreIntegrante} del equipo {nombreEquipo} fue asignado con el ROL {rolIntegrante}"


def normalizar(integrantesLista, nombre):
    integrante = recorrerListaIntegrantes(integrantesLista, nombre)
    if integrante == -1:
        return "El integrante no esta en la lista: "
    else: 
        return integrante.title()


def equipoM(nombreEquipo):
    return nombreEquipo.upper()


def cara(nombreEquipo):
    return len(nombreEquipo)


def siglas(nombreEquipo):
    listaPalabras = nombreEquipo.split()
    sigla = ""
    for palabra in listaPalabras:
        sigla = sigla + palabra[0]
    sigla = sigla.upper()
    return sigla


def unDigito(nombreEquipo):
    val = False
    for caracter in nombreEquipo:
        if caracter.isdigit():
            val = True
    return val


def main():
    nombreEquipo = ""
    comision = ""
    integrantes = []
    rol = []

    nombreEquipoI = input("Ingrese el nombre del equipo: ")
    nombreEquipo = nombreEquipoI
    comisionI = input("Ingrese la comision: ")
    comision = comisionI
    cantidadI = int(input("Ingrese la cantidad de integrantes en el equipo: "))
    while cantidadI <= 0:
        print("Ingresa una cantidad mayor a 0")
        cantidadI = int(input("Ingrese la cantidad de integrantes en el equipo: "))

    for i in range(cantidadI):
        i = i + 1 
        nombreIntegrante = input(f"Ingrese el nombre del integrante del equipo:   integrante n{i}: ")
        rolIntegrante = input(f"Del integrante {nombreIntegrante} n {i} Ingrese su rol: ")
        mensaje = procesarAltaIntegrante(nombreIntegrante, integrantes, rol, rolIntegrante, nombreEquipo)
        print(mensaje)

    print("\nSalida")
    print(f"Equipo: {nombreEquipo} - Comisión: {comision}")
    print("\nTodo")
    todoIntegrante(integrantes)
    todoRol(rol)

    print("\nNormalizar")
    for integrante in integrantes:
        print(normalizar(integrantes, integrante))

    print("\nNombre del equipo en mayuscula")
    print(equipoM(nombreEquipo))

    print(f"\nCantidad de caracteres del equipo {nombreEquipo}")
    print(cara(nombreEquipo))

    print("\nNombre del equipo en siglas")
    print(siglas(nombreEquipo))

    print("\nVerificar si al menos el nombre del equipo tiene un digito")
    if unDigito(nombreEquipo):
        print("Tiene un digito al menos en su nombre")
    else:
        print("No tiene ningun digito en el nombre del equipo")


main()
````

## 9. Registro del avance en GitHub

Para subir el archivo `perfil_equipo.py` al repositorio de forma segura y colaborativa, ejecutamos en la terminal la siguiente secuencia ordenada de comandos:

1. `git pull`: Actualiza la copia local con los últimos cambios que hayan subido los demás compañeros al repositorio remoto.
2. `git status`: Permite verificar que el archivo `codigo/perfil_equipo.py` figure como modificado o nuevo para agregar.
3. `git add codigo/perfil_equipo.py`: Pasa el archivo desarrollado al área de preparación (staging).
4. `git commit -m "Agrega el perfil inicial del equipo"`: Registra los cambios en el historial local con un mensaje claro en presente.
5. `git push`: Envía el commit confirmado al repositorio remoto en GitHub para que esté disponible para todo el equipo.

### Análisis conceptual

* **Inmutabilidad de las cadenas:** Las cadenas de caracteres son inmutables porque una vez creadas en memoria no pueden modificarse posición por posición. Cualquier intento de asignación indexada (como `cadena[0] = 'X'`) produce un error de tipo `TypeError`.
* **Transformar vs. Validar:**
  * **Transformar:** Métodos como `.upper()`, `.lower()`, `.title()` o `.replace()` toman la cadena base y generan una **nueva cadena** con los cambios aplicados, dejando la cadena original intacta.
  * **Validar:** Métodos como `.isalpha()`, `.isdigit()` o `.isalnum()` no generan cadenas nuevas ni modifican el contenido; analizan los caracteres y retornan un valor booleano (`True` o `False`).
