## 3. Problema – Algoritmo – Programa

### a) Análisis del problema (Biblioteca Universitaria)

* **Entradas:**
  * `diasAtraso`: número entero que indica los días transcurridos desde el vencimiento ($\ge 0$).
  * `respuesta`: carácter (`'S'` o `'N'`) que indica si el libro tenía reserva por otro usuario.

* **Procesamiento:**
  1. Validar que `diasAtraso` no sea negativo. Si lo es, mostrar un mensaje de error.
  2. Si `diasAtraso == 0`, la multa asignada es $0.
  3. Si `diasAtraso > 0`:
     * Se cobra la tarifa diaria: $\text{multa} = \text{diasAtraso} \times 800$.
     * Si además `respuesta == 'S'` o `'s'`, se suma el recargo fijo: $\text{multa} = \text{multa} + 2500$.

* **Salida:**
  * `multa`: importe final a pagar por el usuario.

### b) Casos de prueba

| Caso | Días de atraso | Reservado | Resultado Esperado | Justificación / Tipo de caso |
| :---: | :---: | :---: | :--- | :--- |
| **1** | 0 | No (N) | $0 | Límite inferior: devolución a término, sin multa. |
| **2** | 3 | No (N) | $2400 | Caso normal con atraso y sin reserva (3 x 800). |
| **3** | 2 | Sí (S) | $4100 | Caso normal con atraso y reserva (2 x 800 + 2500). |
| **4** | -2 | No (N) | Error / Rechazo | Situación límite inválida (días negativos). |

### Ejercicio 3 (Python)

## 4. Estructuras de datos

* **a) Los datos de una única devolución:**
  * **Estructura elegida:** **Lista simple**.
  * **Justificación:** Una devolución maneja pocos datos agrupados (`[diasAtraso, estaReservado, totalMulta]`). Una lista simple nos permite guardarlos juntos y acceder a cada campo mediante un índice numérico (`0`, `1` o `2`).

* **b) Las multas cobradas durante un día:**
  * **Estructura elegida:** **Lista unidimensional**.
  * **Justificación:** Representa un conjunto secuencial de importes numéricos. Una lista nos permite ir añadiendo datos con `.append()` a medida que devuelven libros y luego recorrerla elemento a elemento con un ciclo `for` para sumar o buscar valores.

* **c) La recaudación de cuatro sucursales durante siete días:**
  * **Estructura elegida:** **Matriz (lista de listas)** de 4 filas y 7 columnas.
  * **Justificación:** Los datos tienen una relación de dos dimensiones fijas: filas para las sucursales y columnas para los días. Se accede con `recaudacion[fila][columna]` y facilita recorrer filas para totales de sucursal o columnas para totales diarios.

## 5. Repaso de programación modular y funciones

### Diferencia entre retornar un valor (`return`) y mostrarlo en pantalla (`print`)

* **`return`:** Es la instrucción que devuelve un valor desde la función hacia la parte del código que la llamó. El dato queda disponible en memoria para guardarlo en una variable, usarlo en cuentas matemáticas o pasarlo como argumento a otra función.
* **`print()`:** Es una salida visual por pantalla orientada al usuario. Muestra información en la consola pero no deja ningún dato disponible en memoria para que el resto del programa siga operando (la función devuelve `None`).


### Ejercicio 5 (Python):

````python
def calcularMulta(diasAtraso, reservado):
    """Calcula y retorna el importe total de la multa."""
    if diasAtraso <= 0:
        total = 0
    else:
        total = diasAtraso * 800
        if reservado:
            total = total + 2500
    return total


def mostrarResultado(nombreUsuario, importe):
    """Muestra en pantalla el importe final asociado a un usuario."""
    print(f"Usuario: {nombreUsuario} | Importe final de la multa: ${importe}")


def main():
    nombreUsuario = input("Ingrese el nombre del usuario: ")
    diasAtraso = int(input("Ingrese los días de atraso: "))

    while diasAtraso < 0:
        print("Error: Los días no pueden ser negativos.")
        diasAtraso = int(input("Ingrese nuevamente los días de atraso: "))

    respuesta = input("¿El libro estaba reservado? (S/N): ")
    estaReservado = (respuesta == "S" or respuesta == "s")

    montoTotal = calcularMulta(diasAtraso, estaReservado)
    mostrarResultado(nombreUsuario, montoTotal)


main()
````

## 6. Repaso de listas

Procesamiento de los importes de 10 usuarios mediante listas unidimensionales. Cada función cumple una sola responsabilidad y realiza los cálculos recorriendo la estructura por posiciones con `range(len(...))`.

### Ejercicio 6 (Python):

````python
def calcularMulta(diasAtraso, reservado):
    """Calcula y retorna el importe total de la multa."""
    if diasAtraso <= 0:
        total = 0
    else:
        total = diasAtraso * 800
        if reservado:
            total = total + 2500
    return total


def registrarMultas(listaMultas, cantidadUsuarios):
    """Solicita los datos de cada usuario, calcula la multa y la agrega a la lista."""
    for i in range(cantidadUsuarios):
        print(f"\n--- Usuario {i + 1} de {cantidadUsuarios} ---")
        diasAtraso = int(input(f"Ingrese días de atraso del usuario {i + 1}: "))
        while diasAtraso < 0:
            diasAtraso = int(input("Error. Ingrese un valor mayor o igual a 0: "))
            
        respuesta = input("¿El libro estaba reservado? (S/N): ")
        estaReservado = (respuesta == "S" or respuesta == "s")
        
        monto = calcularMulta(diasAtraso, estaReservado)
        listaMultas.append(monto)


def mostrarImportes(listaMultas):
    """Muestra todos los importes almacenados en la lista."""
    print("\n--- Registro General de Multas ---")
    for i in range(len(listaMultas)):
        print(f"Usuario {i + 1}: ${listaMultas[i]}")


def obtenerMayorMulta(listaMultas):
    """Retorna la multa de mayor valor encontrada en la lista."""
    if len(listaMultas) == 0:
        return 0
    
    mayor = listaMultas[0]
    for i in range(1, len(listaMultas)):
        if listaMultas[i] > mayor:
            mayor = listaMultas[i]
    return mayor


def calcularPromedio(listaMultas):
    """Calcula y retorna el promedio de los importes."""
    if len(listaMultas) == 0:
        return 0
    
    suma = 0
    for i in range(len(listaMultas)):
        suma = suma + listaMultas[i]
    return suma / len(listaMultas)


def contarSuperanLimite(listaMultas, limite):
    """Cuenta cuántas multas superan un importe recibido por parámetro."""
    contador = 0
    for i in range(len(listaMultas)):
        if listaMultas[i] > limite:
            contador = contador + 1
    return contador


def contarSinMulta(listaMultas):
    """Informa cuántos usuarios devolvieron a tiempo ($0 de multa)."""
    contador = 0
    for i in range(len(listaMultas)):
        if listaMultas[i] == 0:
            contador = contador + 1
    return contador


def main():
    multasRegistradas = []
    totalUsuarios = 10

    print(f"=== Registro de multas para {totalUsuarios} usuarios ===")
    registrarMultas(multasRegistradas, totalUsuarios)

    mostrarImportes(multasRegistradas)

    multaMayor = obtenerMayorMulta(multasRegistradas)
    print(f"\nMulta de mayor importe: ${multaMayor}")

    promedio = calcularPromedio(multasRegistradas)
    print(f"Importe promedio: ${promedio}")

    umbral = 3000
    cantidadSuperan = contarSuperanLimite(multasRegistradas, umbral)
    print(f"Cantidad de multas que superan los ${umbral}: {cantidadSuperan}")

    sinMulta = contarSinMulta(multasRegistradas)
    print(f"Cantidad de usuarios sin multa: {sinMulta}")


main()
````

## 7. Repaso de matrices

### Significado de los índices en `recaudacion[fila][columna]`

* **Primer índice (`fila`):** Representa a la **sucursal** de la biblioteca (va de `0` a `3` para las 4 sucursales).
* **Segundo índice (`columna`):** Representa al **día de la semana** registrado (va de `0` a `6` para los 7 días).

### Ejercicio 7 (Python):

````python
def crearMatriz(cantFilas, cantColumnas):
    """Crea e inicializa una matriz con ceros usando bucles clasicos."""
    matriz = []
    for f in range(cantFilas):
        fila = []
        for c in range(cantColumnas):
            fila.append(0)
        matriz.append(fila)
    return matriz


def cargarMatriz(matriz):
    """Carga la matriz con valores de recaudación de ejemplo."""
    valoresEjemplo = [
        [800, 1600, 0, 2400, 4100, 800, 0],
        [0, 800, 3300, 1600, 0, 4100, 800],
        [2400, 0, 800, 800, 1600, 0, 2500],
        [0, 4100, 2400, 0, 800, 1600, 800]
    ]
    for f in range(len(matriz)):
        for c in range(len(matriz[0])):
            matriz[f][c] = valoresEjemplo[f][c]


def totalPorSucursal(matriz):
    """Informa la recaudación total de cada sucursal."""
    print("\n--- Recaudación Total por Sucursal ---")
    for f in range(len(matriz)):
        sumaSucursal = 0
        for c in range(len(matriz[0])):
            sumaSucursal = sumaSucursal + matriz[f][c]
        print(f"Sucursal {f + 1}: ${sumaSucursal}")


def totalPorDia(matriz):
    """Informa la recaudación total por día y retorna la lista de totales."""
    totalesDias = []
    print("\n--- Recaudación Total por Día ---")
    for c in range(len(matriz[0])):
        sumaDia = 0
        for f in range(len(matriz)):
            sumaDia = sumaDia + matriz[f][c]
        totalesDias.append(sumaDia)
        print(f"Día {c + 1}: ${sumaDia}")
    return totalesDias


def recaudacionGeneral(matriz):
    """Calcula la recaudación acumulada total."""
    totalGeneral = 0
    for f in range(len(matriz)):
        for c in range(len(matriz[0])):
            totalGeneral = totalGeneral + matriz[f][c]
    return totalGeneral


def diaMayorRecaudacion(totalesDias):
    """Busca el número de día (1 a 7) con mayor recaudación total."""
    mayorMonto = totalesDias[0]
    diaMayor = 1
    
    for i in range(1, len(totalesDias)):
        if totalesDias[i] > mayorMonto:
            mayorMonto = totalesDias[i]
            diaMayor = i + 1
            
    return diaMayor, mayorMonto


def main():
    recaudacion = crearMatriz(4, 7)

    cargarMatriz(recaudacion)

    print("Matriz cargada (Filas = Sucursales, Columnas = Días):")
    for f in range(len(recaudacion)):
        print(recaudacion[f])

    totalPorSucursal(recaudacion)
    totalesDiarios = totalPorDia(recaudacion)

    totalGeneral = recaudacionGeneral(recaudacion)
    print(f"\nRecaudación General de la red: ${totalGeneral}")

    diaMax, montoMax = diaMayorRecaudacion(totalesDiarios)
    print(f"El día con mayor recaudación fue el Día {diaMax} con un total de: ${montoMax}")


main()
````
